```python
# ============================================================
# S5 -- CONTEXT-WINDOW STUDENT + KD + artifact-weighted CE
# Student takes a STUDENT_CONTEXT-wide window of zmax/e4 epochs.
# PSG is loaded only to recompute the same n = min(...) used
# during training's KDContextDataset (cheap: labels array only,
# not the full EEG/EOG signal) -- not fed to the student model.
# ============================================================
import os
import sys
sys.modules.pop('Kd_common', None)
import numpy as np
import torch
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, f1_score, cohen_kappa_score

COMMON_PATH = r"D:\22\AA\AA journal\code\student\stat_clinical_significance"

if COMMON_PATH not in sys.path:
    sys.path.insert(0, COMMON_PATH)

from Kd_common import (
    load_3way_split,
    ContextStudentModel,
    LABEL_NAMES,
    save_ensemble_predictions_csv,
    save_per_seed_predictions_csv
)
PSG_EEG_PATH      = r"D:\22\AA\preprocess\preprocessed_FFinal"
STUDENT_DATA_PATH = r"D:\22\AA\preprocess\preprocessed_student_zmax_e4"
SPLIT_PATH        = r"D:\22\AA\AA journal\preprocess\preprocessed_split_v2"

STUDENT_CONTEXT = 7          # MUST match what the checkpoints were trained with
STUDENT_WINDOW  = 2 * STUDENT_CONTEXT + 1

CKPT_DIR = rf"D:\22\AA\AA journal\evaluation\teacher-student\s5_studentC{STUDENT_CONTEXT}_teacherC7_kd"
OUT_DIR  = rf"D:\22\AA\AA journal\evaluation\teacher-student\ensemble_s5_studentC{STUDENT_CONTEXT}"
os.makedirs(OUT_DIR, exist_ok=True)

SEEDS = [42, 123, 256, 789, 999]
D_MODEL = 96
DROPOUT = 0.3
BATCH_SIZE = 32   # matches S5 training (windowed inputs are larger)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")
print(f"Student context: {STUDENT_CONTEXT} (window={STUDENT_WINDOW})")


class SubjectTrackedContextTestDataset(Dataset):
    """Rebuilds the student-side window used by S5's KDContextDataset,
    with the SAME n = min(psg_labels, stu_labels, zmax, e4) subject/
    epoch filtering, but reads only the PSG 'labels' array (not the
    full EEG/EOG signal) since PSG isn't needed for the student's
    forward pass."""
    def __init__(self, subject_list, psg_path, student_path, context=STUDENT_CONTEXT):
        self.context = context
        self.data = []
        self.index = []
        self.subject_ids = []
        for sub in subject_list:
            psg_fp = os.path.join(psg_path, f"{sub}.npz")
            stu_fp = os.path.join(student_path, f"{sub}.npz")
            if not (os.path.exists(psg_fp) and os.path.exists(stu_fp)):
                continue
            with np.load(psg_fp) as d:
                psg_labels = d['labels']
                psg_len = len(psg_labels)
            with np.load(stu_fp) as d:
                zmax_arr = d['zmax_eeg']
                e4_arr = d['e4']
                stu_labels = d['labels'].copy()
            n = min(psg_len, len(stu_labels), zmax_arr.shape[0], e4_arr.shape[0])
            sub_idx = len(self.data)
            self.data.append((zmax_arr[:n], e4_arr[:n], stu_labels[:n], n))
            for i in range(n):
                self.index.append((sub_idx, i))
                self.subject_ids.append(sub)
        print(f"  Subjects: {len(self.data)}   Samples: {len(self.index):,}")

    def __len__(self):
        return len(self.index)

    def __getitem__(self, idx):
        sub_idx, center_i = self.index[idx]
        zmax_arr, e4_arr, labels, n = self.data[sub_idx]

        zmax_window, e4_window = [], []
        for offset in range(-self.context, self.context + 1):
            ei = max(0, min(n - 1, center_i + offset))
            zmax_window.append(zmax_arr[ei])
            e4_window.append(e4_arr[ei])

        zmax_x = np.stack(zmax_window, axis=0)   # (WINDOW, 2, 1920)
        e4_x = np.stack(e4_window, axis=0)       # (WINDOW, 3, 1920)
        y = int(labels[center_i])
        return torch.FloatTensor(zmax_x), torch.FloatTensor(e4_x), torch.tensor(y, dtype=torch.long)


_, _, TEST_SUBS = load_3way_split(SPLIT_PATH)
print(f"Test subjects (from split file): {len(TEST_SUBS)}")

print("Building test dataset (windowed -- uses more RAM than S1/S2/S4)...")
test_ds = SubjectTrackedContextTestDataset(TEST_SUBS, PSG_EEG_PATH, STUDENT_DATA_PATH)
if len(test_ds) == 0:
    raise RuntimeError("test_ds empty -- check PSG_EEG_PATH / STUDENT_DATA_PATH.")

test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

models = []
loaded_seeds = []
print("\nLoading checkpoints...")
for seed in SEEDS:
    ckpt_path = os.path.join(CKPT_DIR, f"best_student_seed{seed}.pt")
    if not os.path.exists(ckpt_path):
        print(f"  WARNING: missing {ckpt_path}")
        continue
    m = ContextStudentModel(d_model=D_MODEL, dropout=DROPOUT, context=STUDENT_CONTEXT).to(device)
    m.load_state_dict(torch.load(ckpt_path, map_location=device))
    m.eval()
    for p in m.parameters():
        p.requires_grad = False
    models.append(m)
    loaded_seeds.append(seed)
    print(f"  Loaded seed {seed}")

if not models:
    raise RuntimeError("No checkpoints found in CKPT_DIR.")
print(f"Total models loaded: {len(models)}")

all_true, all_pred, all_probs = [], [], []
per_seed_preds = {seed: [] for seed in loaded_seeds}
with torch.no_grad():
    for zmax_x, e4_x, y in test_loader:
        zmax_x, e4_x = zmax_x.to(device), e4_x.to(device)
        prob_sum = None
        for seed, m in zip(loaded_seeds, models):
            logits = m(zmax_x, e4_x)
            probs = F.softmax(logits, dim=1)
            per_seed_preds[seed].extend(probs.argmax(dim=1).cpu().numpy())
            prob_sum = probs if prob_sum is None else prob_sum + probs
        avg_probs = (prob_sum / len(models)).cpu().numpy()
        preds = avg_probs.argmax(axis=1)
        all_true.extend(y.numpy())
        all_pred.extend(preds)
        all_probs.extend(avg_probs)

all_true = np.array(all_true)
all_pred = np.array(all_pred)
all_probs = np.array(all_probs)

acc = accuracy_score(all_true, all_pred)
f1 = f1_score(all_true, all_pred, average='macro', labels=[0, 1, 2, 3, 4], zero_division=0)
kappa = cohen_kappa_score(all_true, all_pred, labels=[0, 1, 2, 3, 4])
per_cls = f1_score(all_true, all_pred, average=None, labels=[0, 1, 2, 3, 4], zero_division=0)

print("\n" + "=" * 60)
print(f"S5 ENSEMBLE ({len(models)} seeds)")
print("=" * 60)
print(f"Accuracy : {acc*100:.2f}%")
print(f"Macro F1 : {f1:.4f}")
print(f"Kappa    : {kappa:.4f}")
for name, score in zip(LABEL_NAMES, per_cls):
    print(f"  {name:6s}: {score:.4f}")

out_csv = os.path.join(OUT_DIR, "ensemble_predictions.csv")
save_ensemble_predictions_csv(out_csv, test_ds.subject_ids, all_true, all_pred, all_probs)
print(f"\nSaved: {out_csv}")

per_seed_csv = os.path.join(OUT_DIR, "per_seed_predictions.csv")
save_per_seed_predictions_csv(per_seed_csv, test_ds.subject_ids, per_seed_preds, all_true)
print(f"Saved: {per_seed_csv}")
print("Done.")
```
